Development comparison. The final holdout remains locked. Smoke run: False.

```python
import json
from pathlib import Path

import pandas as pd
import plotly.express as px

result = json.loads(Path("comparison.json").read_text())
assert result["outer_holdout_touched"] is False
print("Smoke run:", result["smoke"])
frame = pd.DataFrame(result["rows"])
frame
```

```python
px.bar(
    frame,
    x="experiment",
    y="delta_vs_control",
    title="Mean fold stability change vs control",
    template="plotly_white",
).show()
```

```python
folds = pd.DataFrame(result["folds"])
px.line(
    folds,
    x="fold",
    y="stability_score",
    color="experiment",
    markers=True,
    template="plotly_white",
    title="Stability by temporal fold",
).show()
```

Only feature blocks change. Early stopping and selection use development folds, so these are not final test scores. ROC AUC and average precision measure ranking; Brier score and log loss assess probabilities.
